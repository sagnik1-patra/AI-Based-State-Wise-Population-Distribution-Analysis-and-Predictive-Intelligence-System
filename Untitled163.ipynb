```python

import os
import json
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ============================================================
# 1. PROJECT PATH
# ============================================================

BASE_PATH = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based State-Wise Population Distribution Analysis"
    r" and Predictive Intelligence System"
)

CSV_PATH = os.path.join(
    BASE_PATH, "RS_Session_259_AU_88_1.csv"
)

H5_PATH = os.path.join(
    BASE_PATH, "population_model.h5"
)

PKL_PATH = os.path.join(
    BASE_PATH, "population_model.pkl"
)

HISTORY_PATH = os.path.join(
    BASE_PATH, "training_history.csv"
)

os.makedirs(BASE_PATH, exist_ok=True)

sns.set_theme(style="whitegrid")

# ============================================================
# 2. LOAD DATASET
# ============================================================

print("\nLoading dataset...")

df = pd.read_csv(CSV_PATH)

df["State/UT"] = df["State/UT"].astype(str).str.strip()

df["Persons"] = pd.to_numeric(
    df["Persons"],
    errors="coerce"
)

df = df.dropna(subset=["State/UT", "Persons"])

# Exclude India aggregate.
df = df[
    df["State/UT"].str.casefold() != "india"
].copy()

df = df.drop_duplicates(subset=["State/UT"])

df = df[df["Persons"] >= 0].copy()

df = df.reset_index(drop=True)

print("Dataset loaded successfully.")
print("Total records:", len(df))

# ============================================================
# 3. LOAD TRAINED MODEL
# ============================================================

print("\nLoading trained models...")

model = tf.keras.models.load_model(
    H5_PATH,
    compile=False
)

components = joblib.load(PKL_PATH)

scaler = components["scaler"]
kmeans = components["kmeans"]
cluster_mapping = components["cluster_mapping"]

print("Models loaded successfully.")

# ============================================================
# 4. PREPROCESS DATA
# ============================================================

df["Log_Persons"] = np.log1p(df["Persons"])

X = df[["Log_Persons"]].values

X_scaled = scaler.transform(X)

# ============================================================
# 5. GENERATE MODEL RECONSTRUCTIONS
# ============================================================

print("\nGenerating model outputs...")

predictions_scaled = model.predict(
    X_scaled,
    verbose=0
)

predictions_log = scaler.inverse_transform(
    predictions_scaled
).flatten()

predictions = np.maximum(
    np.expm1(predictions_log),
    0
)

df["Predicted_Persons"] = np.round(
    predictions
).astype("int64")

df["Actual_Persons"] = df["Persons"].astype("int64")

# ============================================================
# 6. CLUSTER CLASSIFICATION
# ============================================================

clusters = kmeans.predict(X_scaled)

df["Cluster"] = clusters

df["Population_Category"] = [
    cluster_mapping[int(cluster)]
    for cluster in clusters
]

# ============================================================
# 7. CALCULATE PERFORMANCE METRICS
# ============================================================

actual = df["Actual_Persons"].values
predicted = df["Predicted_Persons"].values

mae = mean_absolute_error(actual, predicted)

mse = mean_squared_error(actual, predicted)

rmse = np.sqrt(mse)

r2 = r2_score(actual, predicted)

df["Absolute_Error"] = np.abs(
    actual - predicted
)

df["Percentage_Error"] = np.where(
    actual != 0,
    df["Absolute_Error"] /
    np.maximum(actual, 1) * 100,
    np.nan
)

print("\n========== MODEL PERFORMANCE ==========")
print(f"MAE  : {mae:.4f}")
print(f"MSE  : {mse:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R2   : {r2:.4f}")

print(
    "\nNote: These are in-sample reconstruction "
    "metrics, not forecasting accuracy."
)

# ============================================================
# 8. ACCURACY GRAPH
# ============================================================

print("\nGenerating accuracy graph...")

if os.path.exists(HISTORY_PATH):

    history = pd.read_csv(HISTORY_PATH)

    plt.figure(figsize=(12, 6))

    plt.plot(
        history["Epoch"],
        history["loss"],
        label="Training Loss",
        linewidth=2,
        color="blue"
    )

    if "val_loss" in history.columns:
        plt.plot(
            history["Epoch"],
            history["val_loss"],
            label="Validation Loss",
            linewidth=2,
            color="red"
        )

    plt.title(
        "Neural Network Reconstruction Performance",
        fontsize=15,
        fontweight="bold"
    )

    plt.xlabel("Epoch")
    plt.ylabel("Mean Squared Error (Scaled Values)")

    plt.legend()
    plt.grid(True, alpha=0.3)

    plt.tight_layout()

    plt.savefig(
        os.path.join(BASE_PATH, "accuracy_graph.png"),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

else:
    print(
        "training_history.csv not found. "
        "Creating reconstruction error graph instead."
    )

    sorted_errors = np.sort(
        df["Absolute_Error"].values
    )

    plt.figure(figsize=(12, 6))

    plt.plot(
        range(1, len(sorted_errors) + 1),
        sorted_errors,
        marker="o",
        color="blue"
    )

    plt.title(
        "Reconstruction Absolute Error by Rank"
    )

    plt.xlabel("State/UT Error Rank")
    plt.ylabel("Absolute Error (Persons)")

    plt.tight_layout()

    plt.savefig(
        os.path.join(BASE_PATH, "accuracy_graph.png"),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

# ============================================================
# 9. CORRELATION HEATMAP
# ============================================================

print("\nGenerating heatmap...")

heatmap_columns = [
    "Actual_Persons",
    "Predicted_Persons",
    "Absolute_Error",
    "Percentage_Error"
]

correlation_matrix = df[
    heatmap_columns
].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".3f",
    cmap="coolwarm",
    linewidths=0.5,
    square=True
)

plt.title(
    "Correlation Heatmap of Population Analysis",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    os.path.join(BASE_PATH, "heatmap.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ============================================================
# 10. COMPARISON GRAPH
# ============================================================

print("\nGenerating comparison graph...")

comparison_df = df.sort_values(
    by="Actual_Persons",
    ascending=False
)

x = np.arange(len(comparison_df))

width = 0.4

plt.figure(figsize=(18, 8))

plt.bar(
    x - width / 2,
    comparison_df["Actual_Persons"],
    width=width,
    label="Actual Persons",
    color="royalblue"
)

plt.bar(
    x + width / 2,
    comparison_df["Predicted_Persons"],
    width=width,
    label="Reconstructed Persons",
    color="orange"
)

plt.xticks(
    x,
    comparison_df["State/UT"],
    rotation=90
)

plt.xlabel("States / Union Territories")
plt.ylabel("Number of Persons")

plt.title(
    "Actual vs Reconstructed State-Wise Person Counts",
    fontsize=15,
    fontweight="bold"
)

plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.savefig(
    os.path.join(BASE_PATH, "comparison_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ============================================================
# 11. SAVE RESULT CSV
# ============================================================

print("\nSaving result.csv...")

result_df = df[
    [
        "State/UT",
        "Actual_Persons",
        "Predicted_Persons",
        "Absolute_Error",
        "Percentage_Error",
        "Cluster",
        "Population_Category"
    ]
].copy()

result_df.to_csv(
    os.path.join(BASE_PATH, "result.csv"),
    index=False
)

print("result.csv saved successfully.")

# ============================================================
# 12. RESULT GRAPH
# ============================================================

print("\nGenerating result graph...")

result_sorted = result_df.sort_values(
    by="Actual_Persons",
    ascending=True
)

category_colors = {
    "Low": "green",
    "Medium": "orange",
    "High": "red"
}

colors = [
    category_colors.get(category, "gray")
    for category in result_sorted["Population_Category"]
]

plt.figure(figsize=(14, 12))

plt.barh(
    result_sorted["State/UT"],
    result_sorted["Actual_Persons"],
    color=colors
)

plt.xlabel("Reported Persons")

plt.ylabel("States / Union Territories")

plt.title(
    "State-Wise Population Distribution and Classification",
    fontsize=15,
    fontweight="bold"
)

from matplotlib.patches import Patch

legend_elements = [
    Patch(facecolor="green", label="Low"),
    Patch(facecolor="orange", label="Medium"),
    Patch(facecolor="red", label="High")
]

plt.legend(handles=legend_elements)

plt.grid(axis="x", alpha=0.3)

plt.tight_layout()

plt.savefig(
    os.path.join(BASE_PATH, "result_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ============================================================
# 13. SAVE PREDICTION CSV
# ============================================================

print("\nSaving prediction.csv...")

prediction_df = df[
    [
        "State/UT",
        "Actual_Persons",
        "Predicted_Persons",
        "Population_Category"
    ]
].copy()

prediction_df["Difference"] = (
    prediction_df["Actual_Persons"] -
    prediction_df["Predicted_Persons"]
)

prediction_df.to_csv(
    os.path.join(BASE_PATH, "prediction.csv"),
    index=False
)

print("prediction.csv saved successfully.")

# ============================================================
# 14. PREDICTION GRAPH
# ============================================================

print("\nGenerating prediction graph...")

prediction_sorted = prediction_df.sort_values(
    by="Predicted_Persons",
    ascending=False
)

plt.figure(figsize=(17, 8))

plt.plot(
    prediction_sorted["State/UT"],
    prediction_sorted["Actual_Persons"],
    marker="o",
    linewidth=2,
    label="Actual Persons",
    color="blue"
)

plt.plot(
    prediction_sorted["State/UT"],
    prediction_sorted["Predicted_Persons"],
    marker="s",
    linewidth=2,
    linestyle="--",
    label="Reconstructed Persons",
    color="red"
)

plt.xlabel("States / Union Territories")
plt.ylabel("Number of Persons")

plt.title(
    "State-Wise Neural Network Reconstruction",
    fontsize=15,
    fontweight="bold"
)

plt.xticks(rotation=90)
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()

plt.savefig(
    os.path.join(BASE_PATH, "prediction_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ============================================================
# 15. SAVE METRICS JSON
# ============================================================

metrics = {
    "model": "Neural Network Autoencoder",
    "evaluation_type": "In-sample reconstruction",
    "MAE": float(mae),
    "MSE": float(mse),
    "RMSE": float(rmse),
    "R2_Score": float(r2),
    "total_states": int(len(df)),
    "total_reported_persons": int(
        df["Actual_Persons"].sum()
    )
}

with open(
    os.path.join(BASE_PATH, "evaluation_metrics.json"),
    "w",
    encoding="utf-8"
) as file:
    json.dump(metrics, file, indent=4)

# ============================================================
# 16. FINAL OUTPUT
# ============================================================

print("\n========================================")
print("ALL RESULTS GENERATED SUCCESSFULLY")
print("========================================")

generated_files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "evaluation_metrics.json"
]

for filename in generated_files:
    print("Saved:", filename)

print("\nAll files are located at:")
print(BASE_PATH)

print("\nProject execution completed.")
```
