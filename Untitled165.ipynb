```python

import os
import json
import yaml
import joblib
import random
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    silhouette_score
)

from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.callbacks import EarlyStopping

# ============================================================
# 1. PROJECT CONFIGURATION
# ============================================================

BASE_PATH = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based State-Wise Population Distribution Analysis"
    r" and Predictive Intelligence System"
)

CSV_PATH = os.path.join(
    BASE_PATH,
    "RS_Session_259_AU_88_1.csv"
)

os.makedirs(BASE_PATH, exist_ok=True)

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

sns.set_theme(style="whitegrid")

# PSO PARAMETERS
SWARM_SIZE = 25
MAX_ITERATIONS = 80

INERTIA_WEIGHT = 0.7
COGNITIVE_COEFFICIENT = 1.5
SOCIAL_COEFFICIENT = 1.5

POSITION_NOISE = 0.10
VELOCITY_LIMIT = 0.15

# Neural-network training parameters.
EPOCHS = 150
BATCH_SIZE = 8
LEARNING_RATE = 0.001

# ============================================================
# 2. LOAD DATASET
# ============================================================

print("\nLoading dataset...")

df = pd.read_csv(CSV_PATH)

required_columns = ["State/UT", "Persons"]

for column in required_columns:
    if column not in df.columns:
        raise ValueError(
            f"Required column missing: {column}"
        )

df["State/UT"] = (
    df["State/UT"]
    .astype(str)
    .str.strip()
)

df["Persons"] = pd.to_numeric(
    df["Persons"],
    errors="coerce"
)

df = df.dropna(
    subset=["State/UT", "Persons"]
)

# Remove India aggregate row.
df = df[
    df["State/UT"].str.casefold() != "india"
].copy()

df = df.drop_duplicates(
    subset=["State/UT"]
)

df = df[
    df["Persons"] >= 0
].copy()

df = df.reset_index(drop=True)

if len(df) < 10:
    raise ValueError(
        "Insufficient records for PSO training."
    )

print("Dataset shape:", df.shape)
print("Total states/UTs:", len(df))
print("Total reported persons:", int(df["Persons"].sum()))

# ============================================================
# 3. FEATURE ENGINEERING
# ============================================================

print("\nPreprocessing data...")

df["Log_Persons"] = np.log1p(
    df["Persons"]
)

X = df[
    ["Log_Persons"]
].values.astype(np.float32)

# ============================================================
# 4. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

X_trainval, X_test = train_test_split(
    X,
    test_size=0.20,
    random_state=SEED
)

X_train, X_val = train_test_split(
    X_trainval,
    test_size=0.25,
    random_state=SEED
)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train
).astype(np.float32)

X_val_scaled = scaler.transform(
    X_val
).astype(np.float32)

X_test_scaled = scaler.transform(
    X_test
).astype(np.float32)

X_all_scaled = scaler.transform(
    X
).astype(np.float32)

# ============================================================
# 5. BUILD NEURAL NETWORK AUTOENCODER
# ============================================================

print("\nBuilding neural network...")

inputs = Input(
    shape=(1,),
    name="input_layer"
)

x = Dense(
    16,
    activation="relu",
    name="encoder_1"
)(inputs)

x = Dense(
    8,
    activation="relu",
    name="encoder_2"
)(x)

encoded = Dense(
    2,
    activation="linear",
    name="latent_layer"
)(x)

x = Dense(
    8,
    activation="relu",
    name="decoder_1"
)(encoded)

x = Dense(
    16,
    activation="relu",
    name="decoder_2"
)(x)

outputs = Dense(
    1,
    activation="linear",
    name="output_layer"
)(x)

model = Model(
    inputs=inputs,
    outputs=outputs
)

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=LEARNING_RATE
    ),
    loss="mse",
    metrics=["mae"]
)

model.summary()

# ============================================================
# 6. TRAIN INITIAL NEURAL NETWORK
# ============================================================

print("\nTraining initial neural network...")

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=25,
    restore_best_weights=True
)

history = model.fit(
    X_train_scaled,
    X_train_scaled,
    validation_data=(
        X_val_scaled,
        X_val_scaled
    ),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    callbacks=[early_stopping],
    verbose=1
)

# ============================================================
# 7. CONVERT MODEL WEIGHTS INTO PSO PARTICLES
# ============================================================

print("\nPreparing PSO particles...")

weight_shapes = [
    weight.shape
    for weight in model.get_weights()
]

weight_sizes = [
    int(np.prod(shape))
    for shape in weight_shapes
]

def flatten_weights(weights):

    return np.concatenate([
        weight.flatten()
        for weight in weights
    ]).astype(np.float32)


def restore_weights(vector):

    weights = []
    start = 0

    for shape, size in zip(
        weight_shapes,
        weight_sizes
    ):

        values = vector[
            start:start + size
        ]

        weights.append(
            values.reshape(shape)
        )

        start += size

    return weights


initial_particle = flatten_weights(
    model.get_weights()
)

DIMENSIONS = len(initial_particle)

print("PSO particle dimensions:", DIMENSIONS)

# ============================================================
# 8. NUMPY NEURAL NETWORK FORWARD PASS
# ============================================================

# This avoids repeated TensorFlow predict calls
# during the optimization process.

def forward_pass(particle, X_input):

    weights = restore_weights(particle)

    activation = X_input.astype(np.float32)

    for i in range(0, len(weights), 2):

        kernel = weights[i]
        bias = weights[i + 1]

        activation = (
            activation @ kernel + bias
        )

        if i < len(weights) - 2:

            activation = np.maximum(
                activation,
                0
            )

    return activation


# ============================================================
# 9. PSO FITNESS FUNCTION
# ============================================================

def fitness_function(particle, X_input):

    reconstructed = forward_pass(
        particle,
        X_input
    )

    loss = np.mean(
        (X_input - reconstructed) ** 2
    )

    if not np.isfinite(loss):
        return float("inf")

    return float(loss)


# ============================================================
# 10. INITIALIZE PARTICLE SWARM
# ============================================================

print("\nInitializing PSO swarm...")

rng = np.random.default_rng(SEED)

positions = np.zeros(
    (SWARM_SIZE, DIMENSIONS),
    dtype=np.float32
)

velocities = np.zeros(
    (SWARM_SIZE, DIMENSIONS),
    dtype=np.float32
)

# First particle uses pretrained model weights.
positions[0] = initial_particle.copy()

# Remaining particles use perturbed weights.
for i in range(1, SWARM_SIZE):

    positions[i] = (
        initial_particle +
        rng.normal(
            0,
            POSITION_NOISE,
            DIMENSIONS
        ).astype(np.float32)
    )

# Initialize random velocities.
velocities = rng.uniform(
    -0.05,
    0.05,
    size=(SWARM_SIZE, DIMENSIONS)
).astype(np.float32)

# ============================================================
# 11. INITIALIZE PERSONAL AND GLOBAL BEST
# ============================================================

personal_best_positions = positions.copy()

personal_best_scores = np.array([
    fitness_function(
        particle,
        X_train_scaled
    )
    for particle in positions
])

best_index = int(
    np.argmin(personal_best_scores)
)

global_best_position = (
    personal_best_positions[best_index].copy()
)

global_best_score = float(
    personal_best_scores[best_index]
)

print("Initial best fitness:", global_best_score)

# ============================================================
# 12. PARTICLE SWARM OPTIMIZATION
# ============================================================

print("\nStarting PSO optimization...")

optimization_history = []

for iteration in range(1, MAX_ITERATIONS + 1):

    for i in range(SWARM_SIZE):

        r1 = rng.random(
            DIMENSIONS
        ).astype(np.float32)

        r2 = rng.random(
            DIMENSIONS
        ).astype(np.float32)

        # Velocity update equation.
        cognitive_velocity = (
            COGNITIVE_COEFFICIENT *
            r1 *
            (
                personal_best_positions[i] -
                positions[i]
            )
        )

        social_velocity = (
            SOCIAL_COEFFICIENT *
            r2 *
            (
                global_best_position -
                positions[i]
            )
        )

        velocities[i] = (
            INERTIA_WEIGHT * velocities[i] +
            cognitive_velocity +
            social_velocity
        )

        velocities[i] = np.clip(
            velocities[i],
            -VELOCITY_LIMIT,
            VELOCITY_LIMIT
        )

        # Position update equation.
        positions[i] = (
            positions[i] +
            velocities[i]
        )

        # Evaluate updated particle.
        current_fitness = fitness_function(
            positions[i],
            X_train_scaled
        )

        # Update personal best.
        if current_fitness < personal_best_scores[i]:

            personal_best_scores[i] = (
                current_fitness
            )

            personal_best_positions[i] = (
                positions[i].copy()
            )

            # Update global best.
            if current_fitness < global_best_score:

                global_best_score = (
                    current_fitness
                )

                global_best_position = (
                    positions[i].copy()
                )

    train_loss = fitness_function(
        global_best_position,
        X_train_scaled
    )

    val_loss = fitness_function(
        global_best_position,
        X_val_scaled
    )

    optimization_history.append({
        "Iteration": iteration,
        "Training_Loss": train_loss,
        "Validation_Loss": val_loss,
        "Best_Fitness": global_best_score
    })

    if iteration % 5 == 0 or iteration == 1:

        print(
            f"Iteration {iteration}/{MAX_ITERATIONS} "
            f"| Train Loss: {train_loss:.8f} "
            f"| Validation Loss: {val_loss:.8f}"
        )

# ============================================================
# 13. SELECT FINAL PSO MODEL
# ============================================================

print("\nSelecting final model...")

initial_val_loss = fitness_function(
    initial_particle,
    X_val_scaled
)

pso_val_loss = fitness_function(
    global_best_position,
    X_val_scaled
)

if pso_val_loss <= initial_val_loss:

    final_particle = (
        global_best_position.copy()
    )

    selected_model = "PSO-Optimized Autoencoder"

else:

    final_particle = initial_particle.copy()

    selected_model = "Initial Autoencoder"

print("Selected model:", selected_model)

model.set_weights(
    restore_weights(final_particle)
)

# ============================================================
# 14. EVALUATE ON HELD-OUT TEST DATA
# ============================================================

print("\nEvaluating PSO model...")

test_reconstructed_scaled = model.predict(
    X_test_scaled,
    verbose=0
)

actual_log = scaler.inverse_transform(
    X_test_scaled
).flatten()

reconstructed_log = scaler.inverse_transform(
    test_reconstructed_scaled
).flatten()

actual_test = np.expm1(
    actual_log
)

reconstructed_test = np.maximum(
    np.expm1(reconstructed_log),
    0
)

test_mae = mean_absolute_error(
    actual_test,
    reconstructed_test
)

test_mse = mean_squared_error(
    actual_test,
    reconstructed_test
)

test_rmse = np.sqrt(
    test_mse
)

test_r2 = r2_score(
    actual_test,
    reconstructed_test
)

print("\n========== PSO TEST RESULTS ==========")
print(f"MAE  : {test_mae:.4f}")
print(f"MSE  : {test_mse:.4f}")
print(f"RMSE : {test_rmse:.4f}")
print(f"R2   : {test_r2:.4f}")

# ============================================================
# 15. GENERATE ALL MODEL RECONSTRUCTIONS
# ============================================================

print("\nGenerating state-wise reconstructions...")

all_reconstructed_scaled = model.predict(
    X_all_scaled,
    verbose=0
)

all_reconstructed_log = scaler.inverse_transform(
    all_reconstructed_scaled
).flatten()

all_reconstructed = np.maximum(
    np.expm1(all_reconstructed_log),
    0
)

df["Actual_Persons"] = (
    df["Persons"]
    .round()
    .astype("int64")
)

df["PSO_Predicted_Persons"] = (
    np.round(all_reconstructed)
    .astype("int64")
)

df["Absolute_Error"] = np.abs(
    df["Actual_Persons"] -
    df["PSO_Predicted_Persons"]
)

df["Percentage_Error"] = (
    df["Absolute_Error"] /
    df["Actual_Persons"].replace(0, np.nan)
) * 100

# ============================================================
# 16. K-MEANS CLASSIFICATION
# ============================================================

print("\nClassifying states...")

kmeans = KMeans(
    n_clusters=3,
    random_state=SEED,
    n_init=10
)

df["Cluster"] = kmeans.fit_predict(
    X_all_scaled
)

cluster_order = (
    df.groupby("Cluster")["Persons"]
    .mean()
    .sort_values()
    .index.tolist()
)

cluster_mapping = {
    cluster_order[0]: "Low",
    cluster_order[1]: "Medium",
    cluster_order[2]: "High"
}

df["Population_Category"] = (
    df["Cluster"].map(cluster_mapping)
)

silhouette = silhouette_score(
    X_all_scaled,
    df["Cluster"]
)

# ============================================================
# 17. SAVE PSO H5 MODEL
# ============================================================

h5_path = os.path.join(
    BASE_PATH,
    "pso_population_model.h5"
)

model.save(h5_path)

print("Saved:", h5_path)

# ============================================================
# 18. SAVE PSO PKL COMPONENTS
# ============================================================

pkl_path = os.path.join(
    BASE_PATH,
    "pso_population_model.pkl"
)

joblib.dump(
    {
        "scaler": scaler,
        "kmeans": kmeans,
        "cluster_mapping": cluster_mapping,
        "best_particle": final_particle,
        "weight_shapes": weight_shapes,
        "selected_model": selected_model,
        "algorithm": "Particle Swarm Optimization"
    },
    pkl_path
)

print("Saved:", pkl_path)

# ============================================================
# 19. SAVE PSO YAML CONFIGURATION
# ============================================================

config = {
    "project": (
        "AI-Based State-Wise Population Distribution "
        "Analysis and Predictive Intelligence System"
    ),
    "algorithm": "Particle Swarm Optimization",
    "dataset": os.path.basename(CSV_PATH),
    "seed": SEED,
    "pso_parameters": {
        "swarm_size": SWARM_SIZE,
        "iterations": MAX_ITERATIONS,
        "inertia_weight": INERTIA_WEIGHT,
        "cognitive_coefficient": COGNITIVE_COEFFICIENT,
        "social_coefficient": SOCIAL_COEFFICIENT,
        "position_noise": POSITION_NOISE,
        "velocity_limit": VELOCITY_LIMIT
    },
    "neural_network": {
        "architecture": [1, 16, 8, 2, 8, 16, 1],
        "epochs": EPOCHS,
        "batch_size": BATCH_SIZE,
        "learning_rate": LEARNING_RATE
    },
    "evaluation": {
        "type": "Held-out reconstruction",
        "test_size": 0.20
    }
}

yaml_path = os.path.join(
    BASE_PATH,
    "pso_model_config.yaml"
)

with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as file:
    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

print("Saved:", yaml_path)

# ============================================================
# 20. SAVE PSO MODEL METADATA
# ============================================================

metadata = {
    "created_at": datetime.now().isoformat(),
    "algorithm": "Particle Swarm Optimization",
    "selected_model": selected_model,
    "dataset_records": int(len(df)),
    "total_persons": int(df["Persons"].sum()),
    "best_training_fitness": float(global_best_score),
    "final_validation_loss": float(
        fitness_function(
            final_particle,
            X_val_scaled
        )
    ),
    "test_metrics": {
        "MAE": float(test_mae),
        "MSE": float(test_mse),
        "RMSE": float(test_rmse),
        "R2": float(test_r2)
    },
    "silhouette_score": float(silhouette),
    "limitations": [
        "PSO optimizes reconstruction of observed values.",
        "No future population forecasting is performed.",
        "No temporal or independent predictor data is available."
    ]
}

json_path = os.path.join(
    BASE_PATH,
    "pso_model_metadata.json"
)

with open(
    json_path,
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        metadata,
        file,
        indent=4
    )

print("Saved:", json_path)

# ============================================================
# 21. SAVE PSO OPTIMIZATION HISTORY
# ============================================================

history_df = pd.DataFrame(
    optimization_history
)

history_df.to_csv(
    os.path.join(
        BASE_PATH,
        "pso_optimization_history.csv"
    ),
    index=False
)

print("Saved PSO optimization history.")

# ============================================================
# 22. PSO ACCURACY GRAPH
# ============================================================

print("\nGenerating PSO accuracy graph...")

plt.figure(figsize=(12, 6))

plt.plot(
    history_df["Iteration"],
    history_df["Training_Loss"],
    label="PSO Training Loss",
    linewidth=2,
    color="blue"
)

plt.plot(
    history_df["Iteration"],
    history_df["Validation_Loss"],
    label="PSO Validation Loss",
    linewidth=2,
    color="red"
)

plt.title(
    "PSO Optimization Convergence",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("PSO Iteration")
plt.ylabel("Reconstruction MSE (Scaled Values)")

plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    os.path.join(
        BASE_PATH,
        "pso_accuracy_graph.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ============================================================
# 23. PSO CORRELATION HEATMAP
# ============================================================

print("\nGenerating PSO heatmap...")

heatmap_columns = [
    "Actual_Persons",
    "PSO_Predicted_Persons",
    "Absolute_Error",
    "Percentage_Error"
]

correlation_matrix = df[
    heatmap_columns
].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".3f",
    cmap="coolwarm",
    linewidths=0.5,
    square=True
)

plt.title(
    "PSO Population Analysis Correlation Heatmap",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        BASE_PATH,
        "pso_heatmap.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ============================================================
# 24. PSO COMPARISON GRAPH
# ============================================================

print("\nGenerating PSO comparison graph...")

comparison_df = df.sort_values(
    "Actual_Persons",
    ascending=False
)

x = np.arange(
    len(comparison_df)
)

width = 0.4

plt.figure(figsize=(18, 8))

plt.bar(
    x - width / 2,
    comparison_df["Actual_Persons"],
    width,
    label="Actual Persons",
    color="royalblue"
)

plt.bar(
    x + width / 2,
    comparison_df["PSO_Predicted_Persons"],
    width,
    label="PSO Reconstructed Persons",
    color="orange"
)

plt.xticks(
    x,
    comparison_df["State/UT"],
    rotation=90
)

plt.xlabel("States / Union Territories")
plt.ylabel("Reported Persons")

plt.title(
    "Actual vs PSO Reconstructed Counts",
    fontsize=15,
    fontweight="bold"
)

plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.savefig(
    os.path.join(
        BASE_PATH,
        "pso_comparison_graph.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ============================================================
# 25. SAVE PSO RESULT CSV
# ============================================================

result_columns = [
    "State/UT",
    "Actual_Persons",
    "PSO_Predicted_Persons",
    "Absolute_Error",
    "Percentage_Error",
    "Cluster",
    "Population_Category"
]

result_df = df[
    result_columns
].copy()

result_df.to_csv(
    os.path.join(
        BASE_PATH,
        "pso_result.csv"
    ),
    index=False
)

print("Saved pso_result.csv")

# ============================================================
# 26. PSO RESULT GRAPH
# ============================================================

print("\nGenerating PSO result graph...")

result_sorted = result_df.sort_values(
    "Actual_Persons",
    ascending=True
)

category_colors = {
    "Low": "green",
    "Medium": "orange",
    "High": "red"
}

colors = [
    category_colors.get(
        category,
        "gray"
    )
    for category in result_sorted[
        "Population_Category"
    ]
]

plt.figure(figsize=(14, 12))

plt.barh(
    result_sorted["State/UT"],
    result_sorted["Actual_Persons"],
    color=colors
)

from matplotlib.patches import Patch

legend_elements = [
    Patch(facecolor="green", label="Low"),
    Patch(facecolor="orange", label="Medium"),
    Patch(facecolor="red", label="High")
]

plt.legend(
    handles=legend_elements
)

plt.xlabel("Reported Persons")
plt.ylabel("States / Union Territories")

plt.title(
    "PSO State-Wise Population Classification",
    fontsize=15,
    fontweight="bold"
)

plt.grid(axis="x", alpha=0.3)
plt.tight_layout()

plt.savefig(
    os.path.join(
        BASE_PATH,
        "pso_result_graph.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ============================================================
# 27. SAVE PSO PREDICTION CSV
# ============================================================

prediction_df = df[
    [
        "State/UT",
        "Actual_Persons",
        "PSO_Predicted_Persons",
        "Population_Category"
    ]
].copy()

prediction_df["Difference"] = (
    prediction_df["Actual_Persons"] -
    prediction_df["PSO_Predicted_Persons"]
)

prediction_df.to_csv(
    os.path.join(
        BASE_PATH,
        "pso_prediction.csv"
    ),
    index=False
)

print("Saved pso_prediction.csv")

# ============================================================
# 28. PSO PREDICTION GRAPH
# ============================================================

print("\nGenerating PSO prediction graph...")

prediction_sorted = prediction_df.sort_values(
    "PSO_Predicted_Persons",
    ascending=False
)

plt.figure(figsize=(17, 8))

plt.plot(
    prediction_sorted["State/UT"],
    prediction_sorted["Actual_Persons"],
    marker="o",
    linewidth=2,
    label="Actual Persons",
    color="blue"
)

plt.plot(
    prediction_sorted["State/UT"],
    prediction_sorted["PSO_Predicted_Persons"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="PSO Reconstructed Persons",
    color="red"
)

plt.xlabel("States / Union Territories")
plt.ylabel("Reported Persons")

plt.title(
    "PSO Neural Network Reconstruction Results",
    fontsize=15,
    fontweight="bold"
)

plt.xticks(rotation=90)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    os.path.join(
        BASE_PATH,
        "pso_prediction_graph.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ============================================================
# 29. SAVE PSO EVALUATION METRICS
# ============================================================

metrics = {
    "algorithm": "Particle Swarm Optimization",
    "selected_model": selected_model,
    "evaluation_type": "Held-out reconstruction",
    "MAE": float(test_mae),
    "MSE": float(test_mse),
    "RMSE": float(test_rmse),
    "R2_Score": float(test_r2),
    "Silhouette_Score": float(silhouette),
    "Total_States": int(len(df))
}

with open(
    os.path.join(
        BASE_PATH,
        "pso_evaluation_metrics.json"
    ),
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        metrics,
        file,
        indent=4
    )

# ============================================================
# 30. FINAL SUMMARY
# ============================================================

print("\n============================================")
print("PSO PROJECT EXECUTION COMPLETED")
print("============================================")

generated_files = [
    "pso_population_model.h5",
    "pso_population_model.pkl",
    "pso_model_config.yaml",
    "pso_model_metadata.json",
    "pso_accuracy_graph.png",
    "pso_heatmap.png",
    "pso_comparison_graph.png",
    "pso_result.csv",
    "pso_result_graph.png",
    "pso_prediction.csv",
    "pso_prediction_graph.png",
    "pso_evaluation_metrics.json",
    "pso_optimization_history.csv"
]

for filename in generated_files:
    print("Saved:", filename)

print("\nAll files saved in:")
print(BASE_PATH)

print("\nPSO optimization completed successfully.")
```
