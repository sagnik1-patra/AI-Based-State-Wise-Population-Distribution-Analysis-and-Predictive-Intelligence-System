```python

import os
import json
import yaml
import joblib
import random
import numpy as np
import pandas as pd
import tensorflow as tf

from datetime import datetime
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score,
    silhouette_score
)
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.callbacks import EarlyStopping

# ==========================================================
# 1. PROJECT CONFIGURATION
# ==========================================================

BASE_PATH = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based State-Wise Population Distribution Analysis"
    r" and Predictive Intelligence System"
)

CSV_PATH = os.path.join(
    BASE_PATH, "RS_Session_259_AU_88_1.csv"
)

os.makedirs(BASE_PATH, exist_ok=True)

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# ==========================================================
# 2. LOAD DATASET
# ==========================================================

print("\nLoading dataset...")

df = pd.read_csv(CSV_PATH)

print("\nDataset shape:", df.shape)
print("\nDataset columns:", df.columns.tolist())
print("\nFirst five rows:")
print(df.head())

required_columns = ["State/UT", "Persons"]

for col in required_columns:
    if col not in df.columns:
        raise ValueError(f"Missing required column: {col}")

# ==========================================================
# 3. DATA PREPROCESSING
# ==========================================================

print("\nPreprocessing dataset...")

df["State/UT"] = df["State/UT"].astype(str).str.strip()

df["Persons"] = pd.to_numeric(
    df["Persons"], errors="coerce"
)

df = df.dropna(subset=["State/UT", "Persons"])

# Remove aggregate row to prevent double counting.
india_row = df[
    df["State/UT"].str.casefold() == "india"
].copy()

df = df[
    df["State/UT"].str.casefold() != "india"
].copy()

df = df.drop_duplicates(subset=["State/UT"])

df = df[df["Persons"] >= 0].copy()

df = df.reset_index(drop=True)

if len(df) < 6:
    raise ValueError("Insufficient rows for training.")

print("State/UT records:", len(df))
print("Total reported persons:", int(df["Persons"].sum()))

if not india_row.empty:
    print(
        "India aggregate:",
        int(india_row["Persons"].iloc[0])
    )

# ==========================================================
# 4. FEATURE ENGINEERING
# ==========================================================

# Log transformation reduces the effect of extreme values.
df["Log_Persons"] = np.log1p(df["Persons"])

# This is the only numerical training feature.
X = df[["Log_Persons"]].values

# ==========================================================
# 5. TRAIN-TEST SPLIT
# ==========================================================

X_train, X_test = train_test_split(
    X,
    test_size=0.25,
    random_state=SEED
)

# Fit scaler only on training data.
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
X_all_scaled = scaler.transform(X)

# ==========================================================
# 6. BUILD AUTOENCODER NEURAL NETWORK
# ==========================================================

print("\nBuilding neural network...")

inputs = Input(shape=(1,), name="input_layer")

x = Dense(
    16, activation="relu", name="encoder_1"
)(inputs)

x = Dense(
    8, activation="relu", name="encoder_2"
)(x)

encoded = Dense(
    2, activation="linear", name="latent_layer"
)(x)

x = Dense(
    8, activation="relu", name="decoder_1"
)(encoded)

x = Dense(
    16, activation="relu", name="decoder_2"
)(x)

outputs = Dense(
    1, activation="linear", name="output_layer"
)(x)

model = Model(inputs, outputs)

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="mse",
    metrics=["mae"]
)

model.summary()

# ==========================================================
# 7. TRAIN NEURAL NETWORK
# ==========================================================

print("\nTraining neural network...")

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=30,
    restore_best_weights=True
)

history = model.fit(
    X_train_scaled,
    X_train_scaled,
    validation_data=(
        X_test_scaled,
        X_test_scaled
    ),
    epochs=300,
    batch_size=8,
    callbacks=[early_stopping],
    verbose=1
)

# ==========================================================
# 8. MODEL EVALUATION
# ==========================================================

print("\nEvaluating model...")

test_predictions_scaled = model.predict(
    X_test_scaled,
    verbose=0
)

# Convert reconstructed log counts back to original scale.
actual_log = scaler.inverse_transform(
    X_test_scaled
).flatten()

predicted_log = scaler.inverse_transform(
    test_predictions_scaled
).flatten()

actual_persons = np.expm1(actual_log)

predicted_persons = np.maximum(
    np.expm1(predicted_log), 0
)

mse = mean_squared_error(
    actual_persons, predicted_persons
)

rmse = np.sqrt(mse)

mae = mean_absolute_error(
    actual_persons, predicted_persons
)

r2 = r2_score(
    actual_persons, predicted_persons
)

print("\n========== MODEL RESULTS ==========")
print("MSE :", mse)
print("RMSE:", rmse)
print("MAE :", mae)
print("R2  :", r2)

# These metrics measure reconstruction,
# not future forecasting accuracy.

# ==========================================================
# 9. CLUSTER STATES INTO THREE CATEGORIES
# ==========================================================

print("\nClustering states...")

kmeans = KMeans(
    n_clusters=3,
    random_state=SEED,
    n_init=10
)

clusters = kmeans.fit_predict(X_all_scaled)

df["Cluster"] = clusters

# Assign descriptive labels by ascending cluster mean.
cluster_order = (
    df.groupby("Cluster")["Persons"]
    .mean()
    .sort_values()
    .index.tolist()
)

cluster_mapping = {
    cluster_order[0]: "Low",
    cluster_order[1]: "Medium",
    cluster_order[2]: "High"
}

df["Population_Category"] = (
    df["Cluster"].map(cluster_mapping)
)

silhouette = silhouette_score(
    X_all_scaled, clusters
)

print("\nClustering silhouette score:", silhouette)

# ==========================================================
# 10. GENERATE RECONSTRUCTED VALUES
# ==========================================================

all_predictions_scaled = model.predict(
    X_all_scaled,
    verbose=0
)

all_predictions_log = scaler.inverse_transform(
    all_predictions_scaled
).flatten()

df["Reconstructed_Persons"] = np.maximum(
    np.expm1(all_predictions_log), 0
).round().astype("int64")

df["Absolute_Error"] = (
    df["Persons"] - df["Reconstructed_Persons"]
).abs()

# These are reconstructions of observed counts,
# not predictions of future populations.

# ==========================================================
# 11. SAVE H5 MODEL
# ==========================================================

h5_path = os.path.join(
    BASE_PATH, "population_model.h5"
)

model.save(h5_path)

print("\nSaved H5 model:", h5_path)

# ==========================================================
# 12. SAVE PKL MODEL COMPONENTS
# ==========================================================

pkl_path = os.path.join(
    BASE_PATH, "population_model.pkl"
)

model_components = {
    "scaler": scaler,
    "kmeans": kmeans,
    "cluster_mapping": cluster_mapping,
    "feature_columns": ["Log_Persons"],
    "transformation": "log1p",
    "model_type": "Autoencoder + KMeans"
}

joblib.dump(model_components, pkl_path)

print("Saved PKL components:", pkl_path)

# ==========================================================
# 13. SAVE YAML CONFIGURATION
# ==========================================================

yaml_path = os.path.join(
    BASE_PATH, "model_config.yaml"
)

config = {
    "project": {
        "name": (
            "AI-Based State-Wise Population Distribution "
            "Analysis and Predictive Intelligence System"
        ),
        "dataset": os.path.basename(CSV_PATH),
        "random_seed": SEED
    },
    "data": {
        "target_column": "Persons",
        "state_column": "State/UT",
        "feature_columns": ["Log_Persons"],
        "exclude_aggregate": True
    },
    "model": {
        "type": "Autoencoder",
        "input_features": 1,
        "hidden_layers": [16, 8, 2, 8, 16],
        "activation": "relu",
        "output_activation": "linear",
        "optimizer": "Adam",
        "learning_rate": 0.001,
        "loss": "mse",
        "epochs": 300,
        "batch_size": 8,
        "test_size": 0.25
    },
    "clustering": {
        "algorithm": "KMeans",
        "n_clusters": 3,
        "labels": ["Low", "Medium", "High"]
    },
    "files": {
        "h5": "population_model.h5",
        "pkl": "population_model.pkl",
        "yaml": "model_config.yaml",
        "json": "model_metadata.json"
    }
}

with open(yaml_path, "w", encoding="utf-8") as file:
    yaml.safe_dump(
        config,
        file,
        sort_keys=False,
        allow_unicode=True
    )

print("Saved YAML configuration:", yaml_path)

# ==========================================================
# 14. SAVE JSON METADATA
# ==========================================================

json_path = os.path.join(
    BASE_PATH, "model_metadata.json"
)

metadata = {
    "project_name": config["project"]["name"],
    "created_at": datetime.now().isoformat(),
    "dataset": {
        "filename": os.path.basename(CSV_PATH),
        "state_records": int(len(df)),
        "columns": df.columns.tolist(),
        "total_reported_persons": int(df["Persons"].sum()),
        "minimum_persons": int(df["Persons"].min()),
        "maximum_persons": int(df["Persons"].max()),
        "mean_persons": float(df["Persons"].mean())
    },
    "model": {
        "name": "Neural Network Autoencoder",
        "purpose": "Reconstruction of observed counts",
        "epochs_trained": len(history.history["loss"]),
        "training_loss": float(
            history.history["loss"][-1]
        ),
        "validation_loss": float(
            history.history["val_loss"][-1]
        )
    },
    "evaluation": {
        "mse": float(mse),
        "rmse": float(rmse),
        "mae": float(mae),
        "r2_score": float(r2),
        "metric_scope": "Held-out reconstruction"
    },
    "clustering": {
        "algorithm": "KMeans",
        "number_of_clusters": 3,
        "silhouette_score": float(silhouette),
        "category_counts": {
            str(k): int(v)
            for k, v in df[
                "Population_Category"
            ].value_counts().items()
        }
    },
    "limitations": [
        "The meaning of Persons is not documented.",
        "Only one numerical observation per state is available.",
        "The model does not forecast future population.",
        "Reconstruction scores are not forecasting accuracy."
    ]
}

with open(json_path, "w", encoding="utf-8") as file:
    json.dump(
        metadata,
        file,
        indent=4,
        ensure_ascii=False
    )

print("Saved JSON metadata:", json_path)

# ==========================================================
# 15. SAVE ANALYSIS RESULTS
# ==========================================================

results_path = os.path.join(
    BASE_PATH, "population_analysis_results.csv"
)

df.to_csv(results_path, index=False)

print("Saved analysis CSV:", results_path)

# ==========================================================
# 16. SAVE TRAINING HISTORY
# ==========================================================

history_path = os.path.join(
    BASE_PATH, "training_history.csv"
)

history_df = pd.DataFrame(history.history)

history_df.insert(
    0, "Epoch", range(1, len(history_df) + 1)
)

history_df.to_csv(history_path, index=False)

print("Saved training history:", history_path)

# ==========================================================
# 17. FINAL SUMMARY
# ==========================================================

print("\n==========================================")
print("PROJECT EXECUTION COMPLETED")
print("==========================================")

print("\nGenerated files:")

for filename in [
    "population_model.h5",
    "population_model.pkl",
    "model_config.yaml",
    "model_metadata.json",
    "population_analysis_results.csv",
    "training_history.csv"
]:
    print(" -", filename)

print("\nAll files saved in:")
print(BASE_PATH)
```
