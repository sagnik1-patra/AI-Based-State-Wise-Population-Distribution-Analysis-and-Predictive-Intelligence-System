```python

import os
import json
import yaml
import joblib
import random
import copy
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    silhouette_score
)

from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.callbacks import EarlyStopping

# ============================================================
# 1. PROJECT CONFIGURATION
# ============================================================

BASE_PATH = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based State-Wise Population Distribution Analysis"
    r" and Predictive Intelligence System"
)

CSV_PATH = os.path.join(
    BASE_PATH,
    "RS_Session_259_AU_88_1.csv"
)

os.makedirs(BASE_PATH, exist_ok=True)

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

sns.set_theme(style="whitegrid")

# AIS PARAMETERS
POPULATION_SIZE = 20
GENERATIONS = 60
CLONE_FACTOR = 4
MUTATION_RATE = 0.15
RANDOM_REPLACEMENTS = 3

# Number of top antibodies selected for cloning.
SELECTION_SIZE = 5

# Neural-network training parameters.
EPOCHS = 150
BATCH_SIZE = 8
LEARNING_RATE = 0.001

# AIS optimization is performed on the training data only.
# The test set is held out from model selection.

# ============================================================
# 2. LOAD DATASET
# ============================================================

print("\nLoading dataset...")

df = pd.read_csv(CSV_PATH)

required_columns = ["State/UT", "Persons"]

for col in required_columns:
    if col not in df.columns:
        raise ValueError(
            f"Required column missing: {col}"
        )

df["State/UT"] = (
    df["State/UT"].astype(str).str.strip()
)

df["Persons"] = pd.to_numeric(
    df["Persons"],
    errors="coerce"
)

df = df.dropna(
    subset=["State/UT", "Persons"]
)

# Remove India aggregate to prevent double counting.
df = df[
    df["State/UT"].str.casefold() != "india"
].copy()

df = df.drop_duplicates(
    subset=["State/UT"]
)

df = df[df["Persons"] >= 0].copy()

df = df.reset_index(drop=True)

if len(df) < 10:
    raise ValueError(
        "Insufficient records for AIS training."
    )

print("Dataset shape:", df.shape)
print("Total states/UTs:", len(df))
print("Total persons:", int(df["Persons"].sum()))

# ============================================================
# 3. FEATURE ENGINEERING
# ============================================================

print("\nPreprocessing data...")

df["Log_Persons"] = np.log1p(
    df["Persons"]
)

X = df[["Log_Persons"]].values.astype(
    np.float32
)

# ============================================================
# 4. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

X_trainval, X_test = train_test_split(
    X,
    test_size=0.20,
    random_state=SEED
)

X_train, X_val = train_test_split(
    X_trainval,
    test_size=0.25,
    random_state=SEED
)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train
).astype(np.float32)

X_val_scaled = scaler.transform(
    X_val
).astype(np.float32)

X_test_scaled = scaler.transform(
    X_test
).astype(np.float32)

X_all_scaled = scaler.transform(
    X
).astype(np.float32)

# ============================================================
# 5. BUILD AUTOENCODER MODEL
# ============================================================

print("\nBuilding neural network...")

inputs = Input(shape=(1,))

x = Dense(
    16,
    activation="relu"
)(inputs)

x = Dense(
    8,
    activation="relu"
)(x)

encoded = Dense(
    2,
    activation="linear"
)(x)

x = Dense(
    8,
    activation="relu"
)(encoded)

x = Dense(
    16,
    activation="relu"
)(x)

outputs = Dense(
    1,
    activation="linear"
)(x)

model = Model(
    inputs=inputs,
    outputs=outputs
)

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=LEARNING_RATE
    ),
    loss="mse",
    metrics=["mae"]
)

model.summary()

# ============================================================
# 6. INITIAL NEURAL NETWORK TRAINING
# ============================================================

print("\nTraining initial neural network...")

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=25,
    restore_best_weights=True
)

history = model.fit(
    X_train_scaled,
    X_train_scaled,
    validation_data=(
        X_val_scaled,
        X_val_scaled
    ),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    callbacks=[early_stopping],
    verbose=1
)

# ============================================================
# 7. CONVERT WEIGHTS INTO AIS ANTIBODY
# ============================================================

print("\nPreparing AIS antibodies...")

weight_shapes = [
    weight.shape
    for weight in model.get_weights()
]

weight_sizes = [
    int(np.prod(shape))
    for shape in weight_shapes
]

def flatten_weights(weights):
    return np.concatenate([
        weight.flatten()
        for weight in weights
    ]).astype(np.float32)


def restore_weights(vector):
    weights = []
    start = 0

    for shape, size in zip(
        weight_shapes,
        weight_sizes
    ):
        values = vector[
            start:start + size
        ]

        weights.append(
            values.reshape(shape)
        )

        start += size

    return weights


initial_antibody = flatten_weights(
    model.get_weights()
)

DIMENSIONS = len(initial_antibody)

print("AIS antibody dimensions:", DIMENSIONS)

# ============================================================
# 8. FAST NUMPY FORWARD PASS
# ============================================================

# A NumPy implementation avoids repeatedly invoking
# TensorFlow model.predict during AIS optimization.

def forward_pass(vector, X_input):

    weights = restore_weights(vector)

    a = X_input.astype(np.float32)

    for i in range(0, len(weights), 2):

        kernel = weights[i]
        bias = weights[i + 1]

        a = a @ kernel + bias

        # ReLU for all hidden layers.
        if i < len(weights) - 2:
            a = np.maximum(a, 0)

    return a


# ============================================================
# 9. AIS FITNESS FUNCTION
# ============================================================

def fitness_function(antibody, X_input):

    reconstructed = forward_pass(
        antibody,
        X_input
    )

    mse = np.mean(
        (X_input - reconstructed) ** 2
    )

    if not np.isfinite(mse):
        return float("inf")

    return float(mse)


# ============================================================
# 10. INITIALIZE ANTIBODY POPULATION
# ============================================================

print("\nInitializing AIS population...")

rng = np.random.default_rng(SEED)

population = [
    initial_antibody.copy()
]

for i in range(POPULATION_SIZE - 1):

    noise = rng.normal(
        loc=0,
        scale=0.10,
        size=DIMENSIONS
    ).astype(np.float32)

    antibody = (
        initial_antibody + noise
    )

    population.append(antibody)

# ============================================================
# 11. AIS CLONAL SELECTION ALGORITHM
# ============================================================

print("\nStarting AIS optimization...")

best_antibody = initial_antibody.copy()

best_fitness = fitness_function(
    best_antibody,
    X_train_scaled
)

optimization_history = []

for generation in range(1, GENERATIONS + 1):

    # Evaluate antibodies on training data.
    fitness_values = np.array([
        fitness_function(
            antibody,
            X_train_scaled
        )
        for antibody in population
    ])

    sorted_indices = np.argsort(
        fitness_values
    )

    population = [
        population[i]
        for i in sorted_indices
    ]

    fitness_values = fitness_values[
        sorted_indices
    ]

    # Update global best.
    if fitness_values[0] < best_fitness:

        best_fitness = float(
            fitness_values[0]
        )

        best_antibody = (
            population[0].copy()
        )

    # Select best antibodies.
    selected = population[
        :SELECTION_SIZE
    ]

    clones = []

    # Clone antibodies according to rank.
    for rank, antibody in enumerate(selected):

        number_of_clones = max(
            1,
            CLONE_FACTOR * (
                SELECTION_SIZE - rank
            )
        )

        for clone_index in range(
            number_of_clones
        ):

            clone = antibody.copy()

            # Better antibodies receive smaller mutations.
            mutation_scale = (
                MUTATION_RATE *
                (rank + 1) /
                SELECTION_SIZE
            )

            mutation = rng.normal(
                loc=0,
                scale=mutation_scale,
                size=DIMENSIONS
            ).astype(np.float32)

            clone += mutation

            clones.append(clone)

    # Combine existing antibodies and clones.
    candidates = population + clones

    candidate_fitness = np.array([
        fitness_function(
            antibody,
            X_train_scaled
        )
        for antibody in candidates
    ])

    best_indices = np.argsort(
        candidate_fitness
    )[:POPULATION_SIZE]

    population = [
        candidates[i].copy()
        for i in best_indices
    ]

    # Replace some poor antibodies with diverse solutions.
    for i in range(RANDOM_REPLACEMENTS):

        replacement = (
            best_antibody +
            rng.normal(
                0,
                MUTATION_RATE * 2,
                DIMENSIONS
            ).astype(np.float32)
        )

        population[-(i + 1)] = replacement

    # Record training and validation loss.
    train_loss = fitness_function(
        best_antibody,
        X_train_scaled
    )

    val_loss = fitness_function(
        best_antibody,
        X_val_scaled
    )

    optimization_history.append({
        "Generation": generation,
        "Training_Loss": train_loss,
        "Validation_Loss": val_loss
    })

    if generation % 5 == 0 or generation == 1:

        print(
            f"Generation {generation}/{GENERATIONS} "
            f"| Train Loss: {train_loss:.8f} "
            f"| Validation Loss: {val_loss:.8f}"
        )

# ============================================================
# 12. SELECT BEST AIS MODEL USING VALIDATION DATA
# ============================================================

# The final training-best antibody is available.
# To reduce validation overfitting, we retain the
# initial trained model as a fallback.

initial_weights = flatten_weights(
    model.get_weights()
)

initial_val_loss = fitness_function(
    initial_weights,
    X_val_scaled
)

ais_val_loss = fitness_function(
    best_antibody,
    X_val_scaled
)

if ais_val_loss <= initial_val_loss:

    final_antibody = best_antibody.copy()
    selected_model = "AIS-Optimized"

else:

    final_antibody = initial_weights.copy()
    selected_model = "Initial Autoencoder"

print("\nSelected model:", selected_model)

model.set_weights(
    restore_weights(final_antibody)
)

# ============================================================
# 13. EVALUATE ON HELD-OUT TEST DATA
# ============================================================

print("\nEvaluating AIS model...")

test_reconstructed_scaled = model.predict(
    X_test_scaled,
    verbose=0
)

actual_log = scaler.inverse_transform(
    X_test_scaled
).flatten()

reconstructed_log = scaler.inverse_transform(
    test_reconstructed_scaled
).flatten()

actual_test = np.expm1(actual_log)

reconstructed_test = np.maximum(
    np.expm1(reconstructed_log),
    0
)

test_mae = mean_absolute_error(
    actual_test,
    reconstructed_test
)

test_mse = mean_squared_error(
    actual_test,
    reconstructed_test
)

test_rmse = np.sqrt(test_mse)

test_r2 = r2_score(
    actual_test,
    reconstructed_test
)

print("\n========== AIS TEST RESULTS ==========")
print("MAE :", test_mae)
print("MSE :", test_mse)
print("RMSE:", test_rmse)
print("R2  :", test_r2)

# ============================================================
# 14. GENERATE ALL RECONSTRUCTIONS
# ============================================================

all_reconstructed_scaled = model.predict(
    X_all_scaled,
    verbose=0
)

all_reconstructed_log = scaler.inverse_transform(
    all_reconstructed_scaled
).flatten()

all_reconstructed = np.maximum(
    np.expm1(all_reconstructed_log),
    0
)

df["Actual_Persons"] = (
    df["Persons"].round().astype("int64")
)

df["AIS_Predicted_Persons"] = (
    np.round(all_reconstructed).astype("int64")
)

df["Absolute_Error"] = np.abs(
    df["Actual_Persons"] -
    df["AIS_Predicted_Persons"]
)

df["Percentage_Error"] = (
    df["Absolute_Error"] /
    df["Actual_Persons"].replace(0, np.nan)
) * 100

# ============================================================
# 15. K-MEANS CLUSTERING
# ============================================================

print("\nPerforming state classification...")

kmeans = KMeans(
    n_clusters=3,
    random_state=SEED,
    n_init=10
)

df["Cluster"] = kmeans.fit_predict(
    X_all_scaled
)

cluster_order = (
    df.groupby("Cluster")["Persons"]
    .mean()
    .sort_values()
    .index.tolist()
)

cluster_mapping = {
    cluster_order[0]: "Low",
    cluster_order[1]: "Medium",
    cluster_order[2]: "High"
}

df["Population_Category"] = (
    df["Cluster"].map(cluster_mapping)
)

silhouette = silhouette_score(
    X_all_scaled,
    df["Cluster"]
)

# ============================================================
# 16. SAVE AIS H5 MODEL
# ============================================================

h5_path = os.path.join(
    BASE_PATH,
    "ais_population_model.h5"
)

model.save(h5_path)

print("Saved:", h5_path)

# ============================================================
# 17. SAVE AIS PKL COMPONENTS
# ============================================================

pkl_path = os.path.join(
    BASE_PATH,
    "ais_population_model.pkl"
)

joblib.dump(
    {
        "scaler": scaler,
        "kmeans": kmeans,
        "cluster_mapping": cluster_mapping,
        "best_antibody": final_antibody,
        "weight_shapes": weight_shapes,
        "selected_model": selected_model,
        "algorithm": "AIS-CLONALG"
    },
    pkl_path
)

print("Saved:", pkl_path)

# ============================================================
# 18. SAVE AIS YAML CONFIGURATION
# ============================================================

config = {
    "project": (
        "AI-Based State-Wise Population Distribution "
        "Analysis and Predictive Intelligence System"
    ),
    "algorithm": "Artificial Immune System",
    "optimization_method": "Clonal Selection Algorithm",
    "dataset": os.path.basename(CSV_PATH),
    "seed": SEED,
    "ais_parameters": {
        "population_size": POPULATION_SIZE,
        "generations": GENERATIONS,
        "clone_factor": CLONE_FACTOR,
        "mutation_rate": MUTATION_RATE,
        "random_replacements": RANDOM_REPLACEMENTS,
        "selection_size": SELECTION_SIZE
    },
    "neural_network": {
        "architecture": [1, 16, 8, 2, 8, 16, 1],
        "epochs": EPOCHS,
        "batch_size": BATCH_SIZE,
        "learning_rate": LEARNING_RATE
    },
    "evaluation": {
        "type": "Held-out reconstruction",
        "test_size": 0.20
    }
}

yaml_path = os.path.join(
    BASE_PATH,
    "ais_model_config.yaml"
)

with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as f:
    yaml.safe_dump(
        config,
        f,
        sort_keys=False
    )

# ============================================================
# 19. SAVE AIS METADATA JSON
# ============================================================

metadata = {
    "created_at": datetime.now().isoformat(),
    "algorithm": "AIS-CLONALG",
    "selected_model": selected_model,
    "dataset_records": int(len(df)),
    "total_persons": int(df["Persons"].sum()),
    "best_training_fitness": float(best_fitness),
    "final_validation_loss": float(
        fitness_function(
            final_antibody,
            X_val_scaled
        )
    ),
    "test_metrics": {
        "MAE": float(test_mae),
        "MSE": float(test_mse),
        "RMSE": float(test_rmse),
        "R2": float(test_r2)
    },
    "silhouette_score": float(silhouette),
    "limitations": [
        "AIS optimizes reconstruction, not future forecasts.",
        "No temporal or independent predictor data is available."
    ]
}

json_path = os.path.join(
    BASE_PATH,
    "ais_model_metadata.json"
)

with open(
    json_path,
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        metadata,
        f,
        indent=4
    )

# ============================================================
# 20. SAVE AIS OPTIMIZATION HISTORY
# ============================================================

history_df = pd.DataFrame(
    optimization_history
)

history_df.to_csv(
    os.path.join(
        BASE_PATH,
        "ais_optimization_history.csv"
    ),
    index=False
)

# ============================================================
# 21. AIS ACCURACY GRAPH
# ============================================================

print("\nGenerating AIS accuracy graph...")

plt.figure(figsize=(12, 6))

plt.plot(
    history_df["Generation"],
    history_df["Training_Loss"],
    label="AIS Training Loss",
    linewidth=2,
    color="blue"
)

plt.plot(
    history_df["Generation"],
    history_df["Validation_Loss"],
    label="AIS Validation Loss",
    linewidth=2,
    color="red"
)

plt.title(
    "AIS Optimization Convergence",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("AIS Generation")
plt.ylabel("Reconstruction MSE (Scaled Values)")

plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    os.path.join(
        BASE_PATH,
        "ais_accuracy_graph.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ============================================================
# 22. AIS CORRELATION HEATMAP
# ============================================================

print("\nGenerating AIS heatmap...")

heatmap_columns = [
    "Actual_Persons",
    "AIS_Predicted_Persons",
    "Absolute_Error",
    "Percentage_Error"
]

correlation_matrix = df[
    heatmap_columns
].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".3f",
    cmap="coolwarm",
    linewidths=0.5,
    square=True
)

plt.title(
    "AIS Population Analysis Correlation Heatmap",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        BASE_PATH,
        "ais_heatmap.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ============================================================
# 23. AIS COMPARISON GRAPH
# ============================================================

print("\nGenerating AIS comparison graph...")

comparison_df = df.sort_values(
    "Actual_Persons",
    ascending=False
)

x = np.arange(len(comparison_df))

width = 0.4

plt.figure(figsize=(18, 8))

plt.bar(
    x - width / 2,
    comparison_df["Actual_Persons"],
    width,
    label="Actual Persons",
    color="royalblue"
)

plt.bar(
    x + width / 2,
    comparison_df["AIS_Predicted_Persons"],
    width,
    label="AIS Reconstructed Persons",
    color="orange"
)

plt.xticks(
    x,
    comparison_df["State/UT"],
    rotation=90
)

plt.xlabel("States / Union Territories")
plt.ylabel("Reported Persons")

plt.title(
    "Actual vs AIS Reconstructed Counts",
    fontsize=15,
    fontweight="bold"
)

plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.savefig(
    os.path.join(
        BASE_PATH,
        "ais_comparison_graph.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ============================================================
# 24. SAVE AIS RESULT CSV
# ============================================================

result_columns = [
    "State/UT",
    "Actual_Persons",
    "AIS_Predicted_Persons",
    "Absolute_Error",
    "Percentage_Error",
    "Cluster",
    "Population_Category"
]

result_df = df[
    result_columns
].copy()

result_df.to_csv(
    os.path.join(
        BASE_PATH,
        "ais_result.csv"
    ),
    index=False
)

print("Saved ais_result.csv")

# ============================================================
# 25. AIS RESULT GRAPH
# ============================================================

print("\nGenerating AIS result graph...")

result_sorted = result_df.sort_values(
    "Actual_Persons",
    ascending=True
)

category_colors = {
    "Low": "green",
    "Medium": "orange",
    "High": "red"
}

colors = [
    category_colors.get(
        category,
        "gray"
    )
    for category in result_sorted[
        "Population_Category"
    ]
]

plt.figure(figsize=(14, 12))

plt.barh(
    result_sorted["State/UT"],
    result_sorted["Actual_Persons"],
    color=colors
)

from matplotlib.patches import Patch

legend_elements = [
    Patch(facecolor="green", label="Low"),
    Patch(facecolor="orange", label="Medium"),
    Patch(facecolor="red", label="High")
]

plt.legend(handles=legend_elements)

plt.xlabel("Reported Persons")
plt.ylabel("States / Union Territories")

plt.title(
    "AIS State-Wise Population Classification",
    fontsize=15,
    fontweight="bold"
)

plt.grid(axis="x", alpha=0.3)
plt.tight_layout()

plt.savefig(
    os.path.join(
        BASE_PATH,
        "ais_result_graph.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ============================================================
# 26. SAVE AIS PREDICTION CSV
# ============================================================

prediction_df = df[
    [
        "State/UT",
        "Actual_Persons",
        "AIS_Predicted_Persons",
        "Population_Category"
    ]
].copy()

prediction_df["Difference"] = (
    prediction_df["Actual_Persons"] -
    prediction_df["AIS_Predicted_Persons"]
)

prediction_df.to_csv(
    os.path.join(
        BASE_PATH,
        "ais_prediction.csv"
    ),
    index=False
)

print("Saved ais_prediction.csv")

# ============================================================
# 27. AIS PREDICTION GRAPH
# ============================================================

print("\nGenerating AIS prediction graph...")

prediction_sorted = prediction_df.sort_values(
    "AIS_Predicted_Persons",
    ascending=False
)

plt.figure(figsize=(17, 8))

plt.plot(
    prediction_sorted["State/UT"],
    prediction_sorted["Actual_Persons"],
    marker="o",
    linewidth=2,
    label="Actual Persons",
    color="blue"
)

plt.plot(
    prediction_sorted["State/UT"],
    prediction_sorted["AIS_Predicted_Persons"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="AIS Reconstructed Persons",
    color="red"
)

plt.xlabel("States / Union Territories")
plt.ylabel("Reported Persons")

plt.title(
    "AIS Neural Network Reconstruction Results",
    fontsize=15,
    fontweight="bold"
)

plt.xticks(rotation=90)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    os.path.join(
        BASE_PATH,
        "ais_prediction_graph.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ============================================================
# 28. SAVE AIS EVALUATION METRICS
# ============================================================

metrics = {
    "algorithm": "AIS-CLONALG",
    "selected_model": selected_model,
    "evaluation_type": "Held-out reconstruction",
    "MAE": float(test_mae),
    "MSE": float(test_mse),
    "RMSE": float(test_rmse),
    "R2_Score": float(test_r2),
    "Silhouette_Score": float(silhouette),
    "Total_States": int(len(df))
}

with open(
    os.path.join(
        BASE_PATH,
        "ais_evaluation_metrics.json"
    ),
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        metrics,
        f,
        indent=4
    )

# ============================================================
# 29. FINAL SUMMARY
# ============================================================

print("\n============================================")
print("AIS PROJECT EXECUTION COMPLETED")
print("============================================")

generated_files = [
    "ais_population_model.h5",
    "ais_population_model.pkl",
    "ais_model_config.yaml",
    "ais_model_metadata.json",
    "ais_accuracy_graph.png",
    "ais_heatmap.png",
    "ais_comparison_graph.png",
    "ais_result.csv",
    "ais_result_graph.png",
    "ais_prediction.csv",
    "ais_prediction_graph.png",
    "ais_evaluation_metrics.json",
    "ais_optimization_history.csv"
]

for filename in generated_files:
    print("Saved:", filename)

print("\nAll files saved in:")
print(BASE_PATH)
```
